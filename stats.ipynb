```
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly as py
import math
```

```
p = 0.5
q = 0.5
```

```
arr = []
for i in range(1000+1):
  val =  math.comb(100,i)*(p**i)*(q**(1000-i))
  arr.append(val)
```

```
plt.bar([i for i in range(1001)],arr)
```
